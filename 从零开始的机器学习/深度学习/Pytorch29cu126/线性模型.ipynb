```python
# 导入相关依赖库
import random
import torch

# 1. 生成数据集
def create_dataset(weights, bias, num_example):
    x = torch.normal(0, 1, (num_example, len(weights))) # 生成一个服从均值为0、标准差为1的正态分布的随机张量。生成张量的形状为：num_example行，len(w)列
    y = torch.matmul(x, weights) + bias # 执行矩阵乘法，在 pytorch中进行矩阵乘法的方法是matmul
    y += torch.normal(0, 0.01, y.shape) # 给标签y,加入噪声，模拟数据的误差

    return x, y.reshape((-1, 1)) # 将y转换成列向量
```

```python
true_weights = torch.tensor([2, -3.4])
true_bias = 4.2
features, labels = create_dataset(true_weights,true_bias,1000) # 创建初始化模拟数据集
```
