```python
import torch
from torch import nn
from torch.utils.data import DataLoader
import torchvision
from torch.utils.tensorboard import SummaryWriter
```

```python
dataset = torchvision.datasets.CIFAR10("../Datasets", train=False, download=True, transform=torchvision.transforms.ToTensor())
dataloader = DataLoader(dataset=dataset, batch_size=32)
```

```python
class ReluModel(nn.Module):
    def __init__(self, *args, **kwargs):
        super(ReluModel, self).__init__(*args, **kwargs)
        self.relu1 = nn.ReLU()

    def forward(self, input):
        output = self.relu1(input)
        return output
class SigmoidModel(nn.Module):
    def __init__(self, *args, **kwargs):
        super(SigmoidModel, self).__init__(*args, **kwargs)
        self.sigmoid1 = nn.Sigmoid()

    def forward(self, input):
        output = self.sigmoid1(input)
        return output

writer = SummaryWriter("../logs")
step = 0

model = ReluModel()
sigmodel = SigmoidModel()
for data in dataloader:
    imgs, targets = data
    writer.add_images("ReluInput", imgs, step)
    output1 = model(imgs)
    writer.add_images("ReluOutput", output1, step)
    output2 = sigmodel(imgs)
    writer.add_images("SigmoidOutput", output2, step)
    step += 1

writer.close()
```
