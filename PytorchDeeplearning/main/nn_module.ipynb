```python
import torch
from torch import nn
import torchvision
from torch.utils.tensorboard import SummaryWriter
```

```python
class Test(nn.Module):
    def __init__(self):
        super(Test, self).__init__()

    def forward(self, input):
        output = 3 * input + 1
        return  output

x = torch.tensor(1.0)
model = Test()
output = model(x)
print(output)
```

```python
from torch.utils.data import DataLoader
writer = SummaryWriter("../logs")
step = 0

# 1. torchvision 中使用 dataset 加载数据集
dataset = torchvision.datasets.CIFAR10("../Datasets", train=False, transform=torchvision.transforms.ToTensor(), download=True)

# 2. 使用 torch.utils.data 中的 DataLoader 创建数据集通道
dataloader = DataLoader(dataset=dataset, batch_size=32, shuffle=False)

# 3. 创建卷积神经网络 要求： 只有一个卷积层
class Conv1(nn.Module):
    def __init__(self):
        super(Conv1, self).__init__()
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=6, kernel_size=3, padding=0, stride=1)

    def forward(self, x):
        x = self.conv1(x)
        return x

# 4. 使用卷积神经网络进行训练数据
model = Conv1()
for data in dataloader:
    imgs, targets = data
    output = model(imgs)
    print(imgs.shape)
    print(output.shape)
    output = torch.reshape(output, (-1, 3, 30, 30))
    writer.add_images("Input", imgs, step)
    writer.add_images("Output", output, step)
    step = step + 1

writer.close()
```
