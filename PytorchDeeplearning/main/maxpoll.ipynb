```python
# 导入相关包
import torch
from torch import nn
from torch.utils.tensorboard import SummaryWriter
import torchvision
from torch.utils.data import DataLoader
```

```python
# 导入数据
dataset = torchvision.datasets.CIFAR10("../Datasets", train=False,transform=torchvision.transforms.ToTensor(),
                                       download=True)
```

```python
# 加载数据创建成DataLoader
dataloader = DataLoader(dataset=dataset, batch_size=64, shuffle=False)
```

```python
# 创建单一池化层的神经网络
class OnlyMaxPool(nn.Module):
    def __init__(self, *args, **kwargs):
        super(OnlyMaxPool, self).__init__(*args, **kwargs)
        self.maxpoll1 = nn.MaxPool2d(kernel_size=3, ceil_mode=True)

    def forward(self, input):
        output = self.maxpoll1(input)
        return output

model = OnlyMaxPool()
```

```python
# 循环调用神经网络并使用tensorboard进行展示
writer = SummaryWriter("../logs")
step = 0
for data in dataloader:
    imgs, targets = data
    writer.add_images("OnlyPoolInput", imgs, step)
    output = model(imgs)
    writer.add_images("OnlyPoolOutput", output, step)
    step += 1

writer.close()
```
