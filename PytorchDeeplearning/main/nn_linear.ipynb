```python
import torch
import torchvision
from torch.utils.tensorboard import SummaryWriter
from torch import nn
```

```python
dataset = torchvision.datasets.CIFAR10("../Datasets", train=False, transform=torchvision.transforms.ToTensor(), download=True)

dataloader = torch.utils.data.DataLoader(dataset=dataset, batch_size=64)
```

```python
class LinearModel(nn.Module):
    def __init__(self, *args, **kwargs):
        super(LinearModel, self).__init__(*args, **kwargs)
        self.linear1 = nn.Linear(196608, 10)

    def forward(self, input):
        output = self.linear1(input)
        return output

model = LinearModel()
```

```python
for data in dataloader:
    imgs, targets = data
    print(imgs.shape)
    input = torch.flatten(imgs)
    print(input.shape)
    output = model(input)
    print(output.shape)
```
