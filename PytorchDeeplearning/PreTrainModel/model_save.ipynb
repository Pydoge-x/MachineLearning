```python
import torch
import torchvision

vgg16 = torchvision.models.vgg16()
print(vgg16)
```

# 保存方式1：模型结构 + 模型参数

```python
torch.save(vgg16, "vgg16_method1.pth")
```

# 保存方式2： 模型参数（官方推荐）

```python
torch.save(vgg16.state_dict(), "vgg16_method2.pth")
```

# 加载方式1： 加载模型

```python
model1 = torch.load("vgg16_method1.pth")
print(model1)
```
